# Creating some sample data to input into the database:

```python
from sqlalchemy import create_engine, MetaData, Table

# Create an engine to the local PostgreSQL database
engine = create_engine('postgresql://username:password@localhost:5432/mydatabase')

# Connect to the engine
connection = engine.connect()

# Reflect the tables
metadata = MetaData()
metadata.reflect(bind=engine)

# Print the table names
print(metadata.tables.keys())

# Query a specific table (example: 'mytable')
mytable = Table('mytable', metadata, autoload_with=engine)
result = connection.execute(mytable.select()).fetchall()

# Print the result
for row in result:
    print(row)

# Close the connection
connection.close()
```
